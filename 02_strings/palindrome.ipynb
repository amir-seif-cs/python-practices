# Palindrome

Write a program that checks whether a given string is a palindrome.

A palindrome is a string that reads the same forwards and backwards.

For example:

`madam`

is a palindrome because it remains the same when reversed.


## Concepts

* Strings
* String slicing
* Negative step
* `if` statement
* Comparison operators
* Variables
* `print()`


```python
my_string = input("Enter a string: ")

# Example input: "madam"

if my_string == my_string[::-1]:
    print("The string is a palindrome")
else:
    print("The string is not a palindrome")
```

## Explanation

The program receives a string from the user using `input()`.

The slice `[::-1]` reverses the string by using a negative step.

The original string is compared with its reversed version.

If they are equal, the string is a palindrome. Otherwise, it is not a palindrome.
