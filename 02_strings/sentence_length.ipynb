# Sentence Length

Given the following sentence:

`the quick brown fox jumps over the lazy dog`

Write a program that checks the length of the sentence.

If the sentence contains more than 6 characters, print:

`This sentence is too long`


## Concepts

* Strings
* `len()`
* `if` statement
* Comparison operators
* Variables
* `print()`


```python
my_string = "the quick brown fox jumps over the lazy dog"

if len(my_string) > 6:
    print("This sentence is too long")
```

## Explanation

The sentence is stored in the `my_string` variable.

The `len()` function is used to calculate the number of characters in the string.

The `if` statement checks whether the length of the sentence is greater than 6.

If the condition is true, the message is displayed using `print()`.
