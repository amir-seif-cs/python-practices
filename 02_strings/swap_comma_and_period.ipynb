# Swap Comma and Period

Given the following string:

`32.054,23`

Write a program that swaps the comma and period so that the result becomes:

`32,054.23`


## Concepts

* Strings
* String methods
* String replacement
* Variables
* `print()`


```python
my_string = "32.054,23"

result = my_string.replace(".", "#")
result = result.replace(",", ".")
result = result.replace("#", ",")

print(result)
```

## Explanation

The string is stored in the `my_string` variable.

A temporary character is used to avoid replacing the newly changed characters again.

First, all periods are replaced with `#`. Then, commas are replaced with periods, and finally `#` is replaced with commas.

This swaps the positions of commas and periods in the string.

Finally, the result is displayed using `print()`.
