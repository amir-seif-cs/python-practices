# Delete Character and Word

Given a string, write a program that removes:

* The n-th character from the string.
* The m-th word from the string.

The program should then display the modified string.


## Concepts

* Strings
* String slicing
* `split()`
* `join()`
* Lists
* `pop()`
* Variables


```python
my_string = "the quick brown fox jumps over the lazy dog"

n = int(input("Enter the character position to remove: "))
m = int(input("Enter the word position to remove: "))

# Example input: n = 5, m = 3

my_string = my_string[:n - 1] + my_string[n:]
words = my_string.split()
words.pop(m - 1)

result = " ".join(words)

print(result)
```

## Explanation

The program receives the character position and word position from the user.

String slicing is used to remove the n-th character. Since positions start from 1, `n - 1` is used as the corresponding string index.

The `split()` method converts the string into a list of words. The `pop()` method then removes the m-th word from the list.

Finally, `join()` combines the remaining words back into a string.
