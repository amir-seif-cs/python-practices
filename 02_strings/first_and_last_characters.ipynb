# First and Last Characters

Given the following string:

```text
the quick brown fox jumps over the lazy dog
```

Create a new sentence using the first three characters and the last three characters of the string.


## Concepts

* Strings
* String indexing
* String slicing
* Variables
* String concatenation


````python
my_string = "the quick brown fox jumps over the lazy dog"

result = my_string[:3] + " " + my_string[-3:]

print(result)
````

## Explanation

The string is stored in the `my_string` variable.

The slice `[:3]` extracts the first three characters, while `[-3:]` extracts the last three characters.

The two parts are combined with a space using string concatenation.

Finally, the resulting string is displayed using `print()`.
