# Count Vowels

Write a program that receives a string from the user and counts the number of vowels in the string.

The vowels are:

`a, e, i, o, u`


## Concepts

* User input
* Strings
* `for` loop
* String membership with `in`
* Variables
* `print()`


```python
my_string = input("Enter a string: ")

# Example input: "hello world"

count = 0

for char in my_string.lower():
    if char in "aeiou":
        count += 1

print(f"Number of vowels: {count}")
```

## Explanation

The program receives a string from the user using `input()`.

A `for` loop checks each character in the string. The `lower()` method is used so that uppercase vowels are also counted.

If a character is found in `"aeiou"`, the `count` variable is increased by one.

Finally, the total number of vowels is displayed using an f-string.
