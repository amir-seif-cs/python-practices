# Count "the"

Given the following string:

```text
the quick brown fox jumps over the lazy dog
```

Write a program that counts how many times the word `"the"` appears in the string.


## Concepts

* Strings
* String methods
* `count()`
* Variables
* `print()`


````python
my_string = "the quick brown fox jumps over the lazy dog"

count = my_string.count("the")

print(f'The word "the" appears {count} times.')
````

## Explanation

The string is stored in the `my_string` variable.

The `count()` method counts how many times the specified substring appears in the string.

In this case, `count("the")` returns the number of occurrences of `"the"`.

Finally, the result is displayed using an f-string.
