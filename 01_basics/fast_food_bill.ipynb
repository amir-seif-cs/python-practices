# Fast Food Bill

Write a program that asks the user for the number of pizzas, sandwiches, fries, and drinks.

The program should calculate the total cost of the order, add 10% VAT, and display the final amount in dollars.

Prices:

* Pizza: $5.00

* Sandwich: $3.50

* Fries: $3.00

* Drink: $0.50


## Concepts

* User input
* Variables
* Arithmetic operations
* Percentage calculation
* `print()`


```python
pizza = int(input("Number of pizzas: "))
sandwich = int(input("Number of sandwiches: "))
fries = int(input("Number of fries: "))
drink = int(input("Number of drinks: "))

# Example inputs: 2, 1, 1, 2

total = (pizza * 5) + (sandwich * 3.5) + (fries * 3) + (drink * 0.5)
vat = total * 0.10
final_total = total + vat

print(f"Total: ${final_total:.2f}")
```

## Explanation

The program first receives the number of each item using `input()` and converts the values to integers using `int()`.

The total price is calculated by multiplying the quantity of each item by its price and adding the results together.

A 10% VAT is then calculated and added to the total price.

Finally, an f-string is used to display the final amount in dollars with two decimal places.
