# Tennis Match Time

A device records the total duration of a tennis match in seconds.

Write a program that receives the total number of seconds and converts it into hours, minutes, and seconds.


## Concepts

* User input
* Variables
* Integer division
* Modulo operator (`%`)
* `print()`
,

```python
total_seconds = int(input("Enter the total number of seconds: "))
# Example input: 12283

hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60
seconds = total_seconds % 60

print(f"{hours} hours, {minutes} minutes, {seconds} seconds")
```

## Explanation

The program first receives the total duration in seconds using `input()` and converts it to an integer.

Integer division (`//`) is used to calculate the number of complete hours.

The modulo operator (`%`) gives the remaining seconds after extracting the hours. The remaining seconds are then used to calculate the minutes and seconds.

Finally, the result is displayed using an f-string.
