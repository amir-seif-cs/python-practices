# BMI Calculator

Write a program that asks the user for their height in meters and weight in kilograms.

Calculate the Body Mass Index (BMI) using the following formula:

BMI = weight / height²

The result should be displayed as an integer by removing the decimal part.


## Concepts

* User input
* Variables
* Arithmetic operations
* Type conversion
* Integer conversion
* `print()`


```python
height = float(input("Enter your height in meters: "))
weight = float(input("Enter your weight in kilograms: "))

# Example input: 1.60, 59

bmi = weight / (height ** 2)
bmi = int(bmi)

print(f"BMI: {bmi}")
```

## Explanation

The program receives the height and weight from the user and converts them to floating-point numbers using `float()`.

The BMI is calculated by dividing the weight by the square of the height.

The `int()` function removes the decimal part of the calculated BMI.

Finally, the BMI is displayed using an f-string.
