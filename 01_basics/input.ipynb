# Using `input()`

Practice using Python's `input()` function to receive user input and use it in a program.

## Concepts

- User input
- Variables
- Type conversion
- `print()`

```python
number = int(input("Enter a number: "))
print(number)
```

## Explanation

`input()` returns the user's input as a string.

Since the program expects a number, `int()` converts the input to an integer before storing it in `number`.

The value is then displayed using `print()`.
