# Greeting

Ask the user for their name and print a greeting message.


## Concepts

* User input
* Variables
* Strings
* f-strings
* `print()`


```python
name = input("Enter your name: ")
print(f"hello {name}!")
```

## Explanation

The `input()` function receives the user's name as a string and stores it in the `name` variable.

An f-string is used to insert the value of `name` into the greeting message.

The `print()` function then displays the final greeting.
