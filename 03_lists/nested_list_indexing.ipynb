# Nested List Indexing

Given the following nested list:

```python
list1 = ['a', 'b', ['c', ['d', 'e', ['f', 'g'], 'k'], 'l'], 'm', 'n']
```

Write a program that:

1. Extracts the character `d` from the nested list.
2. Extracts the last two elements as `['m', 'n']`.


## Concepts

* Nested lists
* List indexing
* List slicing
* Variables
* `print()


````python
list1 = ['a', 'b', ['c', ['d', 'e', ['f', 'g'], 'k'], 'l'], 'm', 'n']

d = list1[2][1][0]
last_two = list1[-2:]

print(d)
print(last_two)
````

## Explanation

The nested list is accessed using multiple levels of indexing.

The expression `list1[2][1][0]` accesses the third element, then the second nested list, and finally its first element to extract `d`.

The slice `list1[-2:]` extracts the last two elements of the outer list.

Finally, both results are displayed using `print()`.
